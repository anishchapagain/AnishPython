# Overview of Gradient Descent using Numpy

```python
import numpy as np
```

```python
def gradient_descent(x,y, display=False):
  m_curr = b_curr = 0
  iterations = 500 # trial and find
  n = len(x)
  learning_rate = 0.01 # trial and find

  for i in range(iterations):
    y_predicted = m_curr * x + b_curr

    cost = (1/n) * sum([val**2 for val in (y-y_predicted)]) # Mean Squared Error

    m_derivative = -(2/n) * sum(x * (y - y_predicted)) # partial derivative of m or weight
    b_derivative = -(2/n) * sum(y - y_predicted) # partial derivative of bias

    m_curr = m_curr - (learning_rate * m_derivative)
    b_curr = b_curr - (learning_rate * b_derivative)

    if i%10==0 and display:
      print("m {}, b {}, iteration {} cost:{}".format(m_curr,b_curr,i, cost))

  return m_curr, b_curr
```

```python
x = np.array([1,2,3,4,5,6])
y = np.array([5,7,9,11,13,15])
```

```python
gradient_descent(x,y, True)
```

```python
m_curr, b_curr = gradient_descent(x,y, False)
print(type(m_curr))
print(type(b_curr))
```

```python
# predict
y_pred = m_curr * x + b_curr

print(f"Prediced y {y_pred} when weight {m_curr} and bias {b_curr}.")
```
